## Module 1: Agro-Ecological Zoning (AEZ) Biophysical Framework

### 1.1 Scientific & Agronomic Foundations
The Food and Agriculture Organization (FAO) and the Indian Council of Agricultural Research (ICAR) delineate agricultural geography into distinct **Agro-Ecological Zones (AEZ)**. An AEZ represents a land resource mapping unit defined by climate, landform, soils, and land cover, possessing specific potentials and constraints for land use.

To build an institutional-grade supply chain model, we formalize **5 key environmental dimensions**:
1. **Soil pH**:
   $$\text{pH} = -\log_{10}[H^+]$$
   Governs cation-exchange capacity, micronutrient bioavailability, and microbial enzymatic activity. Optimal range for most field crops is $6.2 - 7.8$. Acidic ($<6.0$) or alkaline/saline ($>8.2$) soils induce severe nutrient lockup.
2. **Soil Organic Carbon (SOC %)**:
   Indicator of soil organic matter (SOM $\approx 1.724 \times$ SOC). Crucial for soil aggregation, biological nutrient cycling, and resilience against soil compaction. High fertility $>0.75\%$; degraded tropical/subtropical soils often drop below $0.40\%$.
3. **Available Water Capacity (AWC in mm/m)**:
   $$\text{AWC} = \text{Field Capacity (FC)} - \text{Permanent Wilting Point (PWP)}$$
   The volume of water held in the root zone between $-33\text{ kPa}$ (matric potential at FC) and $-1500\text{ kPa}$ (PWP). High AWC ($>180\text{ mm/m}$) cushions crops against dry spells; low AWC ($<90\text{ mm/m}$) requires frequent supplemental micro-irrigation.
4. **Mean Annual Precipitation (MAP in mm)**:
   Total climatological annual rainfall, dictating whether production is fundamentally rainfed, supplemental-irrigated, or canal-dependent.
5. **Thermal Heat Unit Window (Cumulative Growing Degree Days - GDD in °C-days)**:
   $$\text{GDD}_{\text{season}} = \sum_{t=1}^{T} \max\left(0, \frac{T_{\max, t} + T_{\min, t}}{2} - T_{\text{base}}\right)$$
   Accumulated thermal energy available during the crop life-cycle. Defines the maturity duration window and sets the ceiling on vegetative biomass accumulation.

## Module 2: PESTLE Quantitative Scoring Framework for Agribusiness

### 2.1 Macroeconomic Architecture
Strategic agribusiness sourcing requires continuous surveillance of external macroeconomic forces. The PESTLE framework is adapted here as a **quantitative multi-attribute scoring model**:
$$\text{Weighted Score} = \sum_{k=1}^{K} w_k \cdot S_k, \quad \sum_{k=1}^{K} w_k = 1.0$$
where:
- $w_k \in [0, 1]$ represents the procurement portfolio impact weight.
- $S_k \in [-5, +5]$ represents the directional force:
  - $+5$: Highly favorable tailwind (margin expansion / supply security).
  - $0$: Neutral / baseline equilibrium.
  - $-5$: Severe headwind (regulatory export ban / catastrophic cost inflation).

### 2.2 Domain Dimensions
- **Political (P)**: Minimum Support Price (MSP) open-ended procurement, geopolitical export bans (e.g., non-basmati rice, onion export tariffs), and inter-state trade barriers.
- **Economic (E)**: Food inflation indices, rural wage growth (MGNREGA wage rate floor), diesel/electricity freight inflation, and interest subvention schemes.
- **Social (S)**: Farmer Producer Organization (FPO) formalization rates, generational farm succession challenges, and dietary protein diversification.
- **Technological (T)**: Adoption of AgriStack digital public infrastructure, remote sensing yield underwriting, automated assaying, and electronic warehouse receipts (e-NWRs).
- **Legal (L)**: Essential Commodities Act (ECA) stock limit enforcement, APMC mandi cess rationalization, and contract farming dispute frameworks.
- **Environmental (E)**: El Niño-Southern Oscillation (ENSO) weather anomalies, groundwater exhaustion, and carbon credit monetization for regenerative agriculture.

## Module 3: Porter's Five Forces Calibrated Industry Attractiveness Model

### 3.1 Structural Forces in Agricultural Value Chains
Michael Porter's Five Forces framework must be re-anchored into agricultural market realities, where biological seasonality, product perishability, and regulatory market committee (APMC) structures redefine competitive dynamics:

1. **Supplier Bargaining Power (Smallholders & Arthiyas)**:
   - Driven by: Farmer fragmentation vs. FPO consolidation; dependence on commission agents (*arthiyas*) for informal credit; availability of alternative local mandis.
2. **Buyer Consolidation (Large Corporates vs. Traditional Millers)**:
   - Driven by: Aggressive procurement desks of FMCG giants (ITC, Adani Wilmar, Cargill, Hindustan Unilever); quick-commerce platforms demanding strict grade standards.
3. **Threat of Substitutes**:
   - Driven by: Import parity pricing (e.g., imported Malaysian palm oil substituting domestic soybean/mustard oil; reconstituted tomato paste substituting fresh tomatoes).
4. **Threat of New Entrants**:
   - Driven by: Capital intensity of cold-chain infrastructure; regulatory licensing requirements in regulated APMC markets; proprietary farmer sourcing network barriers.
5. **Industry Rivalry & Perishability Pressure**:
   - Driven by: Daily biological decay of harvested produce forcing distress sales (*gluts*); working capital velocity; fragmented regional processing capacity.

### 3.2 Industry Attractiveness Index (IAI)
We construct an **Industry Attractiveness Index (IAI)**:
$$\text{Threat Score} = \sum_{j=1}^{5} w_j \cdot T_j, \quad T_j \in [1, 5] \text{ (1 = Low Threat, 5 = Extreme Threat)}$$
$$\text{IAI} = 100 - \left( \frac{\text{Threat Score} - 1.0}{4.0} \times 100 \right)$$
where an IAI of $100$ indicates perfect market attractiveness, and $0$ indicates severe value destruction.


```python
# 3.3 Quantitative Porter's Five Forces Model across Commodity Tiers
porter_weights = {
    "Supplier_Power": 0.25,
    "Buyer_Power": 0.20,
    "Threat_Substitutes": 0.15,
    "Threat_New_Entrants": 0.15,
    "Rivalry_Perishability": 0.25
}

porter_data = [
    {
        "Tier": "High-Perishability Horticulture",
        "Representative_Commodity": "Tomato / Onion",
        "Supplier_Power": 2.2,       # Highly fragmented farmers, desperate to sell at farmgate
        "Buyer_Power": 4.1,          # Quick-commerce & wholesale traders exert extreme price down-pressure
        "Threat_Substitutes": 2.0,   # Fresh culinary staples have low direct culinary substitutes
        "Threat_New_Entrants": 3.8,  # Low barrier for informal traders, but high for cold-chain operators
        "Rivalry_Perishability": 4.9 # Extreme physiological perishability (shelf life 3-7 days)
    },
    {
        "Tier": "Cash Oilseeds",
        "Representative_Commodity": "Soybean / Mustard",
        "Supplier_Power": 3.4,       # Medium holding capacity; farmers hold dry pods for price rallies
        "Buyer_Power": 3.8,          # Large solvent extraction plants and crushing millers dominate
        "Threat_Substitutes": 4.7,   # Extreme threat from cheap imported crude palm oil & sunflower oil
        "Threat_New_Entrants": 2.2,  # Crushing plants require high fixed capex (solvent extraction)
        "Rivalry_Perishability": 2.6 # Low perishability if moisture <10%; can be stored for 6-9 months
    },
    {
        "Tier": "Staple Cereals",
        "Representative_Commodity": "Wheat / Paddy",
        "Supplier_Power": 4.2,       # High political mobilization, assured government MSP safety net
        "Buyer_Power": 3.0,          # Government FCI is dominant buyer; private flour millers compete
        "Threat_Substitutes": 1.8,   # Core dietary caloric foundation; very low demand substitution
        "Threat_New_Entrants": 2.0,  # High regulatory licensing, large silo scale requirements
        "Rivalry_Perishability": 1.9 # Multi-year shelf life under hermetic scientific storage
    }
]

df_porter = pd.DataFrame(porter_data)

# Compute Composite Threat Score & IAI
threat_cols = list(porter_weights.keys())
df_porter["Composite_Threat"] = sum(df_porter[col] * porter_weights[col] for col in threat_cols)
df_porter["Industry_Attractiveness_Index"] = 100.0 - ((df_porter["Composite_Threat"] - 1.0) / 4.0 * 100.0)

display(df_porter[["Tier", "Representative_Commodity", "Composite_Threat", "Industry_Attractiveness_Index"] + threat_cols])

# 3.4 Heatmap Comparison of Porter's Five Forces
plt.figure(figsize=(11, 4.5))
heatmap_data = df_porter.set_index("Representative_Commodity")[threat_cols]
heatmap_data.columns = [c.replace("_", " ") for c in heatmap_data.columns]

sns.heatmap(heatmap_data, annot=True, cmap="YlOrRd", fmt=".1f", linewidths=1.5, cbar_kws={'label': 'Threat Level (1=Low, 5=Severe)'})
plt.title("Porter's Five Forces Threat Intensity Heatmap across Commodity Tiers", weight='bold', pad=15)
plt.ylabel("Target Commodity Tier", weight='bold')
plt.xlabel("Structural Industry Force", weight='bold')
plt.tight_layout()
plt.show()
```

## Module 4: Strategic Produce Segmentation & Positioning Matrix

### 4.1 Portfolio Sourcing Theory
An agribusiness trading and processing enterprise must balance its procurement book across three distinct economic risk profiles:
1. **High-Perishability Horticulture (Tomatoes, Onions, Potatoes - TOP)**:
   - High price volatility, rapid post-harvest degradation, extreme physical perishability.
   - Sourcing focus: Cold-chain velocity, decentralized sorting, forward contract price bands.
2. **Cash Oilseeds (Soybean, Mustard, Groundnut)**:
   - Medium volatility, international tariff linkages, crushing margin dependencies.
   - Sourcing focus: Crushing parity modeling, import parity hedge, moisture-controlled warehousing.
3. **Staple Cereals (Wheat, Maize, Paddy)**:
   - High volume, low volatility, high regulatory presence (MSP floors, open market sales schemes - OMSS).
   - Sourcing focus: Scale logistics, railhead bulk handling, long-term silo storage.

### 4.2 Quantitative Strategic Metrics
- **Addressable Procurement Market Value (INR Crores)**:
  $$\text{Market Value} = \text{Regional Production (Million MT)} \times \text{Modal Price (INR/MT)}$$
- **Supply Volatility Index ($CV_{\text{supply}}$)**:
  $$CV_{\text{supply}} = \frac{\sigma_{\text{daily arrivals}}}{\mu_{\text{daily arrivals}}}$$
- **Price Elasticity of Demand ($E_d$) & Supply ($E_s$)**:
  $$E_d = \frac{\% \Delta Q_d}{\% \Delta P}, \quad E_s = \frac{\% \Delta Q_s}{\% \Delta P}$$
- **Perishability Shelf-Life (Days)**: Physiological threshold before un-refrigerated marketable loss exceeds $15\%$.
